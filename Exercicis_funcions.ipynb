```python
def maxim(a):
    max = a[0]
    for b in range(len(a)):
        if a[b] > max:
            max = a[b]
    return max
print(maxim([12, 3, 4, 43, 55, 60, 56, 23, 12, 2, -76, 60.80, 62]))
```

```python
def primer(n):
    try:
        n = int(n)
        if n > 0:
            primer = True
            for a in range(2, int(n/2)+1):
                if n % a == 0:
                    primer = False
            if primer == True:
                return f"{n} es primer"
            else:
                return f"{n} es compost"
        else:
            return "n no es un nombre natural"
    except:
        return "n no es un nombre natural"
    

print(primer(input("introdueix un nombre natural")))
```

```python
def sumamat(A, B):
     C = []
     for a in range(max([len(A), len(B)])):
        C.append([])
        for b in range(max([len(A[a]), len(B[a])])):
            C[a].append(0)
            
     for a in range(max([len(A), len(B)])):
        for b in range(max([len(A[a]), len(B[a])])):
               if a > len(A) or b > len(A[a]):
                    C[a][b] = B[a][b]
               elif a > len(B) or b > len(B[a]):
                    C[a][b] = B[a][b]
               else:
                    C[a][b] = A[a][b] + B[a][b]
     return C

A = [[1, 0, 1],
     [0, 1, 2],
     [1, 2, 3]]
B = [[0, 2, 1],
     [2, 0, 2],
     [1, 2, 0]]
mostramat(sumamat(A, B))
```

Exercici E1 (codi copiat de producte de matrius.ipynb)

```python
def productemat(A,B):
    C = []
    for a in range(len(A)):
        C.append([])
        for b in range(len(A[a])):
            C[a].append(0)
    if len(A) == len(B[0]) and len(B) == len(A[0]):
        for a in range(len(A)):
            for b in range(len(A[a])):
                for c in range(len(A[a])):
                    C[a][b] += A[a][c]*B[c][b]
        return C
    else:
        print("no multiplicables")
    
A = [[1, 0, 1],
     [0, 1, 2],
     [1, 2, 3]]
B = [[0, 2, 1],
     [2, 0, 2],
     [1, 2, 0]]
mostramat(productemat(A,B))
```

```python
def transposta(A):
    At = []
    for a in range(len(A[0])):
        At.append([])
        for b in range(len(A)):
            At[a].append(0)
    for a in range(len(A[0])):
        for b in range(len(A)):
            At[a][b] = A[b][a]
    return At
A = [[1, 2, 1, 4],
     [0, 1, 2, 1],
     [1, 4, 3, 3]]
mostramat(transposta(A))
```
