```python
def valorsradi(r):
    import math
    radi = r
    perimetre = 2 * math.pi * r
    area = math.pi * r**2
    volum = 4/3 * math.pi * r**3
    valors = {"radi" : radi, "perimetre" : perimetre, "area" : area, "volum" : volum}
    return valors

print(valorsradi(float(input("dona un radi r"))))
```

```python
def interval(a, b, c):
    return (c <= a and c >=b) or (c >= a and c <= b)
print(interval(1, 3, 5))
```

```python
def maxim(a):
    max = a[0]
    for b in range(len(a)-1):
        if a[b] > max:
            max = a[b]
    return max
print(maxim([12, 3, 4, 43, 55, 60, 56, 23, 12, 2, -76]))
```
