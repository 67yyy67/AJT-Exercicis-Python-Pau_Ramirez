```python
def valorsradi(r):
    import math
    radi = r
    perimetre = 2 * math.pi * r
    area = math.pi * r**2
    volum = 4/3 * math.pi * r**3
    valors = {"radi" : radi, "perimetre" : perimetre, "area" : area, "volum" : volum}
    return valors

print(valorsradi(float(input("dona un radi r"))))
```

```python
def interval(a, b, c):
    return (c <= a and c >=b) or (c >= a and c <= b)
print(interval(1, 3, 5))
```

Exercici 3

```python
def maxim(a):
    max = a[0]
    for b in range(len(a)):
        if a[b] > max:
            max = a[b]
    return max
print(maxim([12, 3, 4, 43, 55, 60, 56, 23, 12, 2, -76, 60.80, 62]))
```

Exercici 4

```python
def primer(n):
    try:
        n = int(n)
        if n > 0:
            primer = True
            for a in range(2, int(n/2)+1):
                if n % a == 0:
                    primer = False
            if primer == True:
                return f"{n} es primer"
            else:
                return f"{n} es compost"
        else:
            return "n no es un nombre natural"
    except:
        return "n no es un nombre natural"
    

print(primer(input("introdueix un nombre natural")))
```

```python
def sumamat(A, B):
    C = []
    for a in A:
        for b in A[a]:
            C[a][b] = A[a][b] + B[a][b]

A = [[1, 0, 1],
     [0, 1, 2],
     [1, 2, 3]]
B = [[0, 2, 1],
     [2, 0, 2],
     [1, 2, 0]]
print(sumamat(A, B))
```
