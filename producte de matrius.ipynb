```python
A = [[1, 0, 1],
     [0, 1, 2],
     [1, 2, 3]]
B = [[0, 2, 1],
     [2, 0, 2],
     [1, 2, 0]]
C = []
for a in range(len(A)):
    C.append([])
    for b in range(len(A[a])):
        C[a].append(0)
if len(A) == len(B[0]) and len(B) == len(A[0]):
    for a in range(len(A)):
        for b in range(len(A[a])):
            for c in range(len(A[a])):
                C[a][b] += A[a][c]*B[c][b]
else:
    print("no multiplicables")

print(C)
```
