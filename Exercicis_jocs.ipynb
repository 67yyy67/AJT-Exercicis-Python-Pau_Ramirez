Exercici 1

```python
N = 20
llista_joc = list(range (0 , N))
def decisio_j():
    n = input("vols jugar? (Y/N)")
    if n != "Y" and n != "N":
        print("només Y/N")
        n = decisio_j()
    return n
def demana_index():
    i = input(f"dona un index del 1 al {N}")
    if int(i) not in range(0, N):
        print("index fora de rang")
        i = demana_index()
    return int(i)
def demana_string():
    s = input(f"dona un string")
    return s
def substitució(i, s):
    a = llista_joc
    a[i] = s
    return a
def novallista():
    print(llista_joc)

while True:
    if decisio_j() == "N":
        break
    else:
        pass
    llista_joc = substitució(demana_index(), demana_string())
    novallista()
```

Exercici 2

```python
A = [
    ["-","-","-"],
    ["-","-","-"],
    ["-","-","-"]
    ]
jugador = "X"
def canvijug(a):
    if a == "X":
        b = "O"
    else:
        b = "X"
    return b
def mostramat(A):
    for a in range(len(A)):
        print(A[a])
def posarfitxa(A, jug):
    a = input("a quina fila vols posar fitxa?")
    b = input("a quina columna vols posar fitxa?")
    if ((a or b) not in range(1, 3)) or A[a][b] != "-":
        print("posició no válida")
        return posarfitxa(A, jug)
    else:
        A[a-1][b-1] = jug
while True:
    posarfitxa(A, jugador)
    jugador = canvijug(jugador)
    mostramat(A)
```
