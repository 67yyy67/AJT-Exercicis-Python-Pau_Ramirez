Rectangles

```python
import math

def rectangleE(f, a, b):
    return f(a)*(b-a)
def rectangleD(f, a, b):
    return f(b)*(b-a)
def rectangleM(f, a, b):
    return f((b+a)/2)*(b-a)
def intE(f, a, b, N):
    ret = 0
    for n in range(N):
        ret += rectangleE(f, (a + n*((b-a)/N)), (a + (n+1)*((b-a)/N)))
    return ret
def intD(f, a, b, N):
    ret = 0
    for n in range(N):
        ret += rectangleD(f, (a + n*((b-a)/N)), (a + (n+1)*((b-a)/N)))
    return ret
def intM(f, a, b, N):
    ret = 0
    for n in range(N):
        ret += rectangleM(f, (a + n*((b-a)/N)), (a + (n+1)*((b-a)/N)))
    return ret
def f(x):
    return math.e ** (x ** 3)
a = 0
b = 1
N = 20
print(intE(f, a, b, N))
print(intD(f, a, b, N))
print(intM(f, a, b, N))
```

Trapezi

```python
import math

def Trapezi(f, a, b):
    A1 = rectangleE(f, a, b)
    A2 = (rectangleD(f, a, b) - rectangleE(f, a, b))/2
    return A1 + A2
def intTrapezi(f, a, b, N):
    ret = 0
    for n in range(N):
        ret += Trapezi(f, (a + n*((b-a)/N)), (a + (n+1)*((b-a)/N)))
    return ret
def f(x):
    return math.e ** (x ** 3)
a = 0
b = 1
N = 20
print(intTrapezi(f, a, b, N))
```

Simpson

```python
import math

def Simpson(f, a, b):
    return ((b-a)/6) * (f(a) + 4*f((a+b)/2) + f(b))
def intSimpson(f, a, b, N):
    ret = 0
    for n in range(N):
        ret += Simpson(f, (a + n*((b-a)/N)), (a + (n+1)*((b-a)/N)))
    return ret
def f(x):
    return math.e ** (x ** 3)
a = 0
b = 1
N = 20
print(intSimpson(f, a, b, N))
```

Comparació

```python
def f2(x):
    return x**2
def integralrealxe2(a, b):
    return (b**3)/3 - (a**3)/3
print(intM(f2, 5, 7, 5))
print(intTrapezi(f2, 5, 7, 5))
print(intSimpson(f2, 5, 7, 5))
print(integralrealxe2(5, 7))
llistaN = list(range(5, 51))
for n in llistaN:
    print("-------------------")
    print(n, intM(f2, 5, 7, n))
    print(n, intTrapezi(f2, 5, 7, n))
    print(n, intSimpson(f2, 5, 7, n))
```
