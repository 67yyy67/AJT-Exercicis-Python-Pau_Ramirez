```python
def f(x):
    return x^2
def rectangleE(f, a, b):
    return f(a)*(b-a)
def rectangleD(f, a, b):
    return f(b)*(b-a)
def rectangleM(f, a, b):
    return f(b)*(b-a)
def intE(f, a, b, N):
    ret = 0
    for n in range(N):
        ret += rectangleE(f, (a + n*((b-a)/N)), (a + (n+1)*((b-a)/N)))
    return ret
def intD(f, a, b, N):
    ret = 0
    for n in range(N):
        ret += rectangleD(f, (a + n*((b-a)/N)), (a + (n+1)*((b-a)/N)))
    return ret
def intM(f, a, b, N):
    ret = 0
    for n in range(N):
        ret += rectangleM(f, (a + n*((b-a)/N)), (a + (n+1)*((b-a)/N)))
    return ret
    
```
