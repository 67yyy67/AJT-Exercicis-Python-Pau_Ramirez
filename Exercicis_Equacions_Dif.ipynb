```python
def dibuixar(Xx, Xy):
    Coord = [[[2,-2],  [2,-1], [2,0], [2,1], [2,2]],
             [[1,-2],  [1,-1], [1,0], [1,1], [1,2]],
             [[0,-2],  [0,-1], [0,0], [0,1], [0,2]],
             [[-1,-2],[-1,-1],[-1,0],[-1,1],[-1,2]],
             [[-2,-2],[-2,-1],[-2,0],[-2,1],[-2,2]]]
    #= [[] for a in range(int(input("files")))]
    #b = input("columnes")
    #for a in Coord:
    #    Coord[a] = [[] for a in range(int(b))]
    #    for b in Coord[a]:
    #        Coord[a][b] = [ - int(len(Coord[a])/2) + b + 1, - int(len(Coord)/2) + a + 1]

    XCoord = Coord
    for a in range(len(Coord)):
        for b in range(len(Coord[a])):
            XCoord[a][b] = [Coord[a][b][1] + Xx(Coord[a][b][1], Coord[a][b][0]),Coord[a][b][0] + Xy(Coord[a][b][1],  Coord[a][b][0] )]
    return XCoord

def X1x(x, y):
    return -y
def X1y(x, y):
    return x
def Euler(x, y, h, T):
    for a in range(0, int(T/h)):
        x = x - h*X1x(x, y)
        y = y + h*X1y(x, y)
    return x, y
print(dibuixar(X1x, X1y))   
x0  = 0
y0 = 1
h = 0.01
T = 10
print(Euler(x0, y0, h, T))
```

```python
def EulerX1(x, y, h, T):
    for a in range(0, int(T/h)):
        x = x - (h/2)*(X1x(x, y) + X1x(x+h, y+h))
        y = y + (h/2)*(X1y(x, y) + X1y(x+h, y+h))
    return x, y
x0  = 0
y0 = 1
h = 0.01
T = 10
print(EulerX1(x0, y0, h, T))
```
