Exercici 1

```python
#Que vol dir "dibuixar"?????
def X1x(x, y):
    return -y
def X1y(x, y):
    return x
def Euler(x, y, h, T):
    for a in range(0, int(T/h)):
        x = x - h*X1x(x, y)
        y = y + h*X1y(x, y)
    return x, y
x0  = 0
y0 = 1
h = 0.01
T = 10
print(Euler(x0, y0, h, T))
```

```python
def EulerX1(x, y, h, T):
    for a in range(0, int(T/h)):
        x = x - (h/2)*(X1x(x, y) + X1x(x+h, y+h))
        y = y + (h/2)*(X1y(x, y) + X1y(x+h, y+h))
    return x, y
x0  = 0
y0 = 1
h = 0.01
T = 10
print(EulerX1(x0, y0, h, T))
```
