```python
class Punt2D:
    def __init__(self, x, y):
        try:
            self.x = float(x)
            self.y = float(y)
        except:
            print("X i Y han de ser nombres")
    @property
    def distancia_origen(self):
        print((self.x**2 + self.y**2)**0.5)
    def Distancia(self, p):
        try:
            px = p.x
            py = p.y
            print(((px - self.x )**2 + (py - self.y)**2)**0.5)
        except:
            print(f"{p} no está correctament definit com un punt")
    def Mou(self, dx, dy):
        self.x += dx
        self.y += dy
    @property
    def __str__(self):
        print(self.x , self.y)

p = Punt2D(2, 2)
d = Punt2D(4, 4)
d.Distancia(p)
d.__str__
```

```python
class CompteBancari:
    def __init__(self, titular, saldo = 0):
        if type(titular) != str:
            raise ValueError("Titualr ha de ser un string")
        self.titular = titular
        try:
            self.saldo = float(saldo)
            if self.saldo < 0:
                raise ValueError("El saldo ha de ser positiu")
        except:
            raise ValueError("El saldo ha de ser un float")
    def Ingresa(self, q):
        if (type(q) != float and type(q) != int) or q <= 0:
            raise ValueError("Q ha de ser positiu")
        self.saldo = self.saldo + q
    def Retira(self, q):
        if (type(q) != float and type(q) != int) or (q <= 0):
            raise ValueError("Q ha de ser positiu")
        elif q > self.saldo:
            raise ValueError("No tens prou saldo")
        self.saldo = self.saldo - q
    def Transfereix(self, compte, q):
        if (type(q) != float and type(q) != int) or (q <= 0):
            raise ValueError("Q ha de ser positiu")
        elif q > self.saldo:
            raise ValueError("No tens prou saldo")
        if not isinstance(compte, CompteBancari):
            raise ValueError("el compte no existeix")
        self.saldo = self.saldo - q
        compte.saldo = compte.saldo + q
juan = CompteBancari("juan", 1000)
roberto = CompteBancari("roberto", 1000)
print(juan.saldo, roberto.saldo)
juan.Ingresa(100)
print(juan.saldo, roberto.saldo)
juan.Retira(100)
print(juan.saldo, roberto.saldo)
roberto.Transfereix(juan, 100)
print(juan.saldo, roberto.saldo)
```

```python
class Rectangle:
    def __init__(self, b, h):
        try:
            self.b = float(b)
            self.h = float(h)
        except:
            raise ValueError("b i h han de ser floats")
        if b <= 0 or h <= 0:
            raise ValueError("b i h hsn de ser positius")
    def A(self)
        return self.b * self.h
    def P(self)
        return 2 * self.b + 2 * self.h
    @property
    def diagonal(self):
        return (self.b ** 2 + self.h ** 2) ** (1/2)
```
