```python
class Punt2D:
    def __init__(self, x, y):
        try:
            self.x = float(x)
            self.y = float(y)
        except:
            print("X i Y han de ser nombres")
    @property
    def distancia_origen(self):
        print((self.x**2 + self.y**2)**0.5)
    def Distancia(self, p):
        try:
            px = p.x
            py = p.y
            print(((px - self.x )**2 + (py - self.y)**2)**0.5)
        except:
            print(f"{p} no está correctament definit com un punt")
    def Mou(self, dx, dy):
        self.x += dx
        self.y += dy
    @property
    def __str__(self):
        print(self.x , self.y)

p = Punt2D(2, 2)
d = Punt2D(4, 4)
d.Distancia(p)
d.__str__
```

```python
class CompteBancari:
    def __init__(self, titular, saldo = 0):
        self.titular = titular
        self.saldo = saldo
        try:
            self.saldo = float(saldo)
            if self.saldo < 0:
                print("saldo ha de ser positiu")
        except:
            print("saldo ha de ser un float")
    def Ingresa(self, q):
        if (type(q) != float and type(q) != int) or q < 0:
            print("q ha de ser un nombre positiu")
            break
        self.saldo = self.saldo + q
    def Retira(self, q):
        if (type(q) != float and type(q) != int) or (q < 0):
            print("q ha de ser un nombre positiu")
            break
        elif q > self.saldo:
            print("no tens prou saldo")
            break
        self.saldo = self.saldo - q
```
